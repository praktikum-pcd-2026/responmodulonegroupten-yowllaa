# FLIPPING CITRA

Kode di bawah digunakan untuk melakukan proses **flip** atau pembalikan citra secara horizontal dan vertikal menggunakan operasi manipulasi piksel secara manual. Program diawali dengan pembuatan fungsi bernama `fungsi()` yang memiliki dua parameter, yaitu `img` sebagai citra input dan `mode` untuk menentukan jenis pembalikan citra.

Pada bagian awal fungsi, ukuran citra diambil menggunakan `img.shape[:2]` untuk mendapatkan tinggi (`h`) dan lebar (`w`) gambar. Selanjutnya, program membuat matriks kosong menggunakan `np.zeros_like(img)` yang memiliki ukuran sama dengan citra asli dan digunakan untuk menyimpan hasil pembalikan gambar.

Proses pembalikan dilakukan menggunakan dua buah perulangan `for` untuk mengakses setiap piksel pada citra. Jika parameter `mode='X'`, maka program melakukan flip horizontal dengan menukar posisi piksel berdasarkan sumbu X sehingga gambar terlihat terbalik ke samping. Sedangkan jika `mode='Y'`, program melakukan flip vertikal dengan menukar posisi piksel berdasarkan sumbu Y sehingga gambar terlihat terbalik atas-bawah.

Setelah fungsi selesai dibuat, program menerapkan proses flip pada setiap citra grayscale. Citra `DinoA2` dan `DinoB2` dibalik secara vertikal menggunakan mode `'Y'`, sedangkan `DinoC2` dan `DinoD2` dibalik secara horizontal menggunakan mode `'X'`. Hasil pembalikan kemudian disimpan pada variabel `flipA`, `flipB`, `flipC`, dan `flipD`.

Selanjutnya, hasil setiap citra ditampilkan menggunakan `plt.subplot()` dalam susunan 2×2. Fungsi `plt.imshow()` digunakan untuk menampilkan gambar grayscale dengan parameter `cmap='gray'`, sedangkan `plt.title()` digunakan untuk memberi label pada masing-masing gambar. Selain itu, `plt.axis('off')` digunakan untuk menghilangkan sumbu koordinat agar tampilan visualisasi menjadi lebih rapi dan mudah diamati.

````python
def fungsi(img, mode='X'):
    h, w = img.shape[:2]
    hasil = np.zeros_like(img)

    for i in range(h):
        for j in range(w):
            if mode == 'X':      # horizontal
                hasil[i, j] = img[i, w-1-j]
            elif mode == 'Y':    # vertical
                hasil[i, j] = img[h-1-i, j]

    return hasil
flipB = fungsi(DinoB2, 'Y')
flipA = fungsi(DinoA2, 'Y')
flipC = fungsi(DinoC2, 'X')
flipD = fungsi(DinoD2, 'X')

plt.subplot(2, 2, 1)
plt.imshow(flipA, cmap="gray")
plt.title('A')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(flipB, cmap="gray")
plt.title('B')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(flipC, cmap="gray")
plt.title('C')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(flipD, cmap="gray")
plt.title('D')
plt.axis('off')
````

# MELAKUKAN ROTASI

Kode di bawah digunakan untuk melakukan proses **rotasi citra** secara manual dengan sudut tertentu menggunakan manipulasi posisi piksel. Program diawali dengan pembuatan fungsi bernama `fungsi_rotasi()` yang memiliki dua parameter, yaitu `image` sebagai citra input dan `val` sebagai nilai sudut rotasi.

Pada awal fungsi, ukuran citra diambil menggunakan `image.shape` untuk mendapatkan tinggi (`h`) dan lebar (`w`) gambar. Nilai sudut rotasi kemudian diubah menjadi bilangan bulat dan dibatasi menggunakan operasi modulo `% 360` agar sudut tetap berada dalam rentang 0–359 derajat.

Program kemudian melakukan pengecekan terhadap nilai sudut rotasi. Jika sudut bernilai `0`, maka citra dikembalikan tanpa perubahan. Jika sudut bernilai `90`, program membuat matriks baru dengan ukuran yang ditukar antara tinggi dan lebar citra, kemudian setiap piksel dipindahkan ke posisi baru sehingga menghasilkan rotasi 90 derajat searah jarum jam.

Pada kondisi `180`, program membuat matriks dengan ukuran yang sama seperti citra asli, lalu memindahkan setiap piksel ke posisi berlawanan sehingga menghasilkan rotasi 180 derajat. Sedangkan pada kondisi `270`, program kembali membuat matriks baru dan memindahkan posisi piksel sehingga menghasilkan rotasi 270 derajat atau 90 derajat berlawanan arah jarum jam.

Setelah fungsi selesai dibuat, program menerapkan rotasi pada citra hasil flip sebelumnya. Citra `flipA` dan `flipC` diputar sebesar 270 derajat, sedangkan `flipB` dan `flipD` diputar sebesar 90 derajat. Hasil rotasi kemudian disimpan pada variabel `rotasiA`, `rotasiB`, `rotasiC`, dan `rotasiD`.

Selanjutnya, seluruh hasil citra ditampilkan menggunakan `plt.subplot()` dalam susunan 2×2. Fungsi `plt.imshow()` digunakan untuk menampilkan citra grayscale dengan parameter `cmap='gray'`, sedangkan `plt.title()` digunakan untuk memberikan label A, B, C, dan D pada masing-masing gambar sehingga hasil rotasi dapat diamati dengan lebih mudah.

````python
def fungsi_rotasi(image, val):
    h, w = image.shape
    val = int(val) % 360
    if val == 0: return image
    elif val == 90:
        rot = np.zeros((w, h), dtype=image.dtype)
        for i in range(h):
            for j in range(w): rot[j, h - 1 - i] = image[i, j]
        return rot
    elif val == 180:
        rot = np.zeros_like(image, dtype=image.dtype)
        for i in range(h):
            for j in range(w): rot[h - 1 - i, w - 1 - j] = image[i, j]
        return rot
    elif val == 270:
        rot = np.zeros((w, h), dtype=image.dtype)
        for i in range(h):
            for j in range(w): rot[w - 1 - j, i] = image[i, j]
        return rot
    return image

rotasiA = fungsi_rotasi(flipA, 270)
rotasiB = fungsi_rotasi(flipB, 90)
rotasiC = fungsi_rotasi(flipC,270)
rotasiD = fungsi_rotasi(flipD, 90)

plt.subplot(2, 2, 1)
plt.imshow(rotasiA, cmap="gray")
plt.title('A')

plt.subplot(2, 2, 2)
plt.imshow(rotasiB, cmap="gray")
plt.title('B')

plt.subplot(2, 2, 3)
plt.imshow(rotasiC, cmap="gray")
plt.title('C')

plt.subplot(2, 2, 4)
plt.imshow(rotasiD, cmap="gray")
plt.title('D')
````

# TRANSLASI

Kode di atas digunakan untuk melakukan proses **translasi citra** atau pergeseran posisi gambar menggunakan manipulasi koordinat piksel secara manual. Program diawali dengan pembuatan fungsi bernama `translasi()` yang memiliki tiga parameter, yaitu `image` sebagai citra input, `tx` sebagai nilai pergeseran horizontal, dan `ty` sebagai nilai pergeseran vertikal.

Pada bagian awal fungsi, ukuran citra diambil menggunakan `image.shape` untuk memperoleh tinggi (`h`) dan lebar (`w`) gambar. Selanjutnya, program membuat matriks kosong menggunakan `np.zeros()` dengan ukuran yang sama seperti citra asli. Matriks ini digunakan untuk menyimpan hasil translasi citra.

Proses translasi dilakukan menggunakan dua buah perulangan `for` untuk mengakses setiap piksel pada citra. Untuk setiap piksel, program menghitung posisi baru menggunakan persamaan:
- `new_x = x + tx`
- `new_y = y + ty`

Nilai `tx` digunakan untuk menggeser gambar ke kanan atau kiri, sedangkan `ty` digunakan untuk menggeser gambar ke atas atau bawah. Setelah posisi baru dihitung, program melakukan pengecekan batas agar koordinat baru tetap berada di dalam ukuran citra. Jika posisi masih valid, maka nilai piksel dari citra asli dipindahkan ke posisi baru pada matriks hasil.

Setelah fungsi translasi selesai dibuat, program menerapkan translasi pada citra hasil rotasi sebelumnya. Citra `rotasiA` digeser ke kanan dan bawah sebesar `(50, 51)`, citra `rotasiB` digeser ke kiri dan bawah sebesar `(-51, 52)`, citra `rotasiC` digeser ke kanan dan atas sebesar `(50, -52)`, sedangkan citra `rotasiD` digeser ke kiri dan atas sebesar `(-50, -51)`.

Hasil translasi kemudian disimpan pada variabel `translasiA`, `translasiB`, `translasiC`, dan `translasiD`. Selanjutnya, seluruh citra hasil translasi ditampilkan menggunakan `plt.subplot()` dalam susunan 2×2 dengan bantuan `plt.imshow()` dan parameter `cmap='gray'` agar gambar tampil dalam bentuk grayscale sehingga perubahan posisi citra dapat diamati dengan jelas.

````python
def translasi(image, tx, ty):
    h, w = image.shape
    hasil = np.zeros((h, w), dtype=image.dtype)

    for y in range(h):
        for x in range(w):
            new_x = x + tx
            new_y = y + ty

            if 0 <= new_x < w and 0 <= new_y < h:
                hasil[new_y][new_x] = image[y][x]
    return hasil

translasiA = translasi(rotasiA, 50, 51)
translasiB = translasi(rotasiB, -51, 52)
translasiC = translasi(rotasiC, 50, -52)
translasiD = translasi(rotasiD,-50, -51)

plt.subplot(2, 2, 1)
plt.imshow(translasiA, cmap="gray")

plt.subplot(2, 2, 2)
plt.imshow(translasiB, cmap="gray")


plt.subplot(2, 2, 3)
plt.imshow(translasiC, cmap="gray")


plt.subplot(2, 2, 4)
plt.imshow(translasiD, cmap="gray")

````

# DILATASI

Kode di bawah digunakan untuk melakukan proses **dilatasi citra** atau pembesaran ukuran gambar dengan metode manipulasi piksel secara manual. Program diawali dengan pembuatan fungsi bernama `dilatasi()` yang memiliki parameter `image` sebagai citra input, serta parameter tambahan `p1`, `p2`, `p3`, dan `p4` yang digunakan untuk mengatur pemotongan area hasil dilatasi.

Pada bagian awal fungsi, ukuran citra diambil menggunakan `image.shape` untuk memperoleh tinggi (`h`) dan lebar (`w`) gambar. Selanjutnya, program membuat matriks kosong bernama `dilated` dengan ukuran dua kali lebih besar dari citra asli, yaitu `(h * 2, w * 2)`. Matriks ini digunakan untuk menyimpan hasil pembesaran citra.

Proses dilatasi dilakukan menggunakan dua buah perulangan `for` untuk mengakses setiap piksel pada matriks hasil. Nilai piksel pada citra asli kemudian disalin ke posisi baru menggunakan:
- `image[int(y / 2)][int(x / 2)]`

Metode ini menyebabkan setiap piksel pada citra asli direplikasi menjadi area yang lebih besar sehingga ukuran gambar tampak diperbesar atau mengalami efek zoom.

Setelah proses pembesaran selesai, program melakukan pemotongan area citra menggunakan parameter `p1`, `p2`, `p3`, dan `p4`. Variabel:
- `sy` dan `ey` digunakan untuk menentukan batas atas dan bawah citra,
- `sx` dan `ex` digunakan untuk menentukan batas kiri dan kanan citra.

Proses ini bertujuan untuk mengambil bagian tertentu dari hasil dilatasi sehingga posisi gambar dapat disesuaikan sesuai kebutuhan.

Selanjutnya, fungsi dilatasi diterapkan pada citra hasil translasi sebelumnya. Setiap citra diberikan parameter pemotongan yang berbeda agar posisi hasil pembesaran sesuai dengan susunan objek yang diinginkan. Hasil dilatasi kemudian disimpan pada variabel `dilatasiA`, `dilatasiB`, `dilatasiC`, dan `dilatasiD`.

Pada bagian akhir, seluruh hasil citra ditampilkan menggunakan `plt.subplot()` dalam susunan 2×2. Fungsi `plt.imshow()` digunakan untuk menampilkan citra grayscale dengan parameter `cmap='gray'`, sehingga hasil pembesaran citra dapat diamati dengan lebih jelas.

````python
def dilatasi(image, p1=0, p2=0, p3=0, p4=0):
    h, w = image.shape
    dilated = np.zeros((h * 2, w * 2)).astype(int)
    for y in range(h * 2):
        for x in range(w * 2):

            dilated[y][x] = image[int(y / 2)][int(x / 2)]
    sy, ey = max(0, p1), min(h * 2, h * 2 - p2)
    sx, ex = max(0, p3), min(w * 2, w * 2 - p4)
    return dilated[sy:ey, sx:ex]


dilatasiA = dilatasi(translasiA, 200, 0, 200, 0)
dilatasiB = dilatasi(translasiB, 200, 0, 0, 200)
dilatasiC = dilatasi(translasiC, 0, 200, 200, 0)
dilatasiD = dilatasi(translasiD, 0, 200, 0, 200)

plt.subplot(2, 2, 1)
plt.imshow(dilatasiA, cmap="gray")

plt.subplot(2, 2, 2)
plt.imshow(dilatasiB, cmap="gray")


plt.subplot(2, 2, 3)
plt.imshow(dilatasiC, cmap="gray")


plt.subplot(2, 2, 4)
plt.imshow(dilatasiD, cmap="gray")
````

# MENGGABUNGKAN CITRA

Kode di bawah digunakan untuk melakukan proses **penggabungan citra** secara horizontal dan vertikal menggunakan manipulasi matriks dengan bantuan library **NumPy**. Program terdiri dari dua fungsi utama, yaitu `fungsi_dua_b()` untuk menggabungkan citra secara menyamping (horizontal) dan `fungsi_dua_c()` untuk menggabungkan citra secara bertumpuk (vertikal).

Pada fungsi `fungsi_dua_b()`, dua citra input diubah terlebih dahulu menjadi array NumPy menggunakan `np.array()`. Selanjutnya, program menentukan ukuran tinggi maksimum dari kedua citra serta menghitung total lebar gabungan kedua gambar. Setelah itu, dibuat matriks kosong menggunakan `np.zeros()` untuk menampung hasil penggabungan citra.

Citra pertama ditempatkan di sisi kiri matriks hasil, sedangkan citra kedua ditempatkan di sisi kanan menggunakan teknik indexing array. Dengan cara ini, kedua gambar dapat digabungkan secara horizontal dalam satu frame.

Selanjutnya, fungsi `fungsi_dua_c()` digunakan untuk menggabungkan dua citra secara vertikal. Pada fungsi ini, tinggi hasil diperoleh dari penjumlahan tinggi kedua citra, sedangkan lebar diambil dari ukuran lebar terbesar di antara kedua gambar. Program kemudian membuat matriks kosong dan menempatkan citra pertama di bagian atas serta citra kedua di bagian bawah.

Setelah kedua fungsi selesai dibuat, program mulai menggabungkan citra hasil dilatasi sebelumnya. Citra `dilatasiA` dan `dilatasiB` digabungkan secara horizontal dan disimpan pada variabel `mergeA`. Kemudian, citra `dilatasiC` dan `dilatasiD` juga digabungkan secara horizontal dan disimpan pada variabel `mergeB`.

Selanjutnya, hasil `mergeA` dan `mergeB` digabungkan kembali secara vertikal menggunakan fungsi `fungsi_dua_c()` sehingga menghasilkan satu citra utuh yang disimpan pada variabel `merge`.

Pada bagian akhir, fungsi `plt.imshow()` digunakan untuk menampilkan hasil akhir penggabungan citra dalam bentuk grayscale dengan parameter `cmap='gray'`. Dengan proses ini, seluruh potongan gambar dapat disusun kembali menjadi satu tampilan citra lengkap.

````python
def fungsi_dua_b(citra_1, citra_2):
    c1, c2 = np.array(citra_1), np.array(citra_2)
    tinggi = max(c1.shape[0], c2.shape[0])
    lebar = c1.shape[1] + c2.shape[1]
    hasil = np.zeros((tinggi, lebar)).astype(int)
    hasil[0:c1.shape[0], 0:c1.shape[1]] = c1
    hasil[0:c2.shape[0], c1.shape[1]:] = c2
    return hasil
def fungsi_dua_c(citra_1, citra_2):
    c1, c2 = np.array(citra_1), np.array(citra_2)
    tinggi = c1.shape[0] + c2.shape[0]
    lebar = max(c1.shape[1], c2.shape[1])
    hasil = np.zeros((tinggi, lebar)).astype(int)
    hasil[0:c1.shape[0], 0:c1.shape[1]] = c1
    hasil[c1.shape[0]:, 0:c2.shape[1]] = c2
    return hasil

mergeA = fungsi_dua_b(dilatasiA, dilatasiB)
mergeB = fungsi_dua_b(dilatasiC, dilatasiD)
merge = fungsi_dua_c(mergeA, mergeB)

plt.imshow(merge, cmap='gray')
````

# MENGGABUNGKAN NAILONG DENGAN ATRIBUT TAMBAHAN

Kode di bawah digunakan untuk melakukan proses **resize** dan **overlay citra** sehingga beberapa objek gambar dapat digabungkan menjadi satu komposisi citra baru. Program diawali dengan pembuatan fungsi `resize()` yang digunakan untuk mengubah ukuran citra secara manual berdasarkan tinggi (`target_h`) dan lebar (`target_w`) yang diinginkan.

Pada fungsi `resize()`, program terlebih dahulu mengambil ukuran asli citra menggunakan `image.shape`. Selanjutnya dibuat matriks kosong menggunakan `np.zeros()` untuk menampung hasil perubahan ukuran. Proses resize dilakukan menggunakan dua buah perulangan `for` untuk memetakan setiap piksel pada citra target ke posisi piksel pada citra asli dengan perhitungan skala:
- `src_y = int(y * h / target_h)`
- `src_x = int(x * w / target_w)`

Metode ini membuat ukuran citra dapat diperbesar atau diperkecil sesuai kebutuhan.

Selanjutnya, program membuat fungsi `overlay()` yang digunakan untuk menempelkan citra lain di atas citra dasar (`base`). Pada fungsi ini, gambar overlay terlebih dahulu diubah ukurannya menggunakan fungsi `resize()`. Kemudian program membuat salinan citra dasar menggunakan `base.copy()` agar citra asli tidak berubah.

Proses overlay dilakukan dengan memeriksa setiap piksel pada citra overlay. Posisi baru piksel dihitung menggunakan parameter `x_pos` dan `y_pos` sehingga gambar dapat ditempatkan pada lokasi tertentu. Program juga melakukan pengecekan batas agar piksel tidak keluar dari ukuran citra dasar.

Selain itu, digunakan kondisi:
```python
if img_resized[y][x] > 20:

````python
def resize(image, target_h, target_w):

    h, w = image.shape

    hasil = np.zeros((target_h, target_w), dtype=image.dtype)

    for y in range(target_h):
        for x in range(target_w):

            src_y = int(y * h / target_h)
            src_x = int(x * w / target_w)

            hasil[y][x] = image[src_y][src_x]

    return hasil


def overlay(base, overlay_img,
                    x_pos, y_pos,
                    target_w, target_h):

    img_resized = resize(
        overlay_img,
        target_h,
        target_w
    )

    result = base.copy()

    h_base, w_base = result.shape

    for y in range(target_h):
        for x in range(target_w):

            ny = y + y_pos
            nx = x + x_pos

            if 0 <= ny < h_base and 0 <= nx < w_base:

                if img_resized[y][x] > 20:

                    result[ny][nx] = img_resized[y][x]

    return result

nailong_sultan = merge.copy()

Money_crop = Money[50:750, 50:950]

nailong_sultan = overlay(
    base        = nailong_sultan,
    overlay_img = Money_crop,

    x_pos       = 0,
    y_pos       = 0,

    target_w    = 1800,
    target_h    = 800
)

Money_crop = Money[50:750, 50:950]

nailong_sultan = overlay(
    base        = nailong_sultan,
    overlay_img = Money_crop,

    x_pos       = 0,
    y_pos       = 0,

    target_w    = 1800,
    target_h    = 800
)

Kacamata_crop = Kacamata[80:520, 50:950]

nailong_sultan = overlay(
    base        = nailong_sultan,
    overlay_img = Kacamata_crop,

    x_pos       = -40,

    y_pos       = 145,

    target_w    = 1900,
    target_h    = 780
)

Udud_crop = Udud[250:950, 200:950]

nailong_sultan = overlay(
    base        = nailong_sultan,
    overlay_img = Udud_crop,

    x_pos       = 540,

    y_pos       = 610,

    target_w    = 950,
    target_h    = 950
)

plt.figure(figsize=(10,12))

plt.imshow(nailong_sultan, cmap='gray')

plt.title(
    "Sang Sultan Nailong",
    fontsize=18,
    fontweight='bold'
)

plt.axis('off')

plt.show()
````

# HASIL HISTOGRAM

Kode di bawah digunakan untuk membuat dan membandingkan **histogram citra** antara citra hasil penggabungan awal (`merge`) dan citra akhir (`nailong_sultan`). Histogram pada pengolahan citra berfungsi untuk menunjukkan distribusi intensitas piksel dari suatu gambar, mulai dari nilai gelap (0) hingga terang (255).

Pada bagian awal, program membuat dua array kosong menggunakan `np.zeros(256, dtype=int)` yaitu `hist_asli` dan `hist_gabungan`. Array tersebut memiliki panjang 256 karena citra grayscale memiliki rentang intensitas piksel dari 0 sampai 255. Kedua array digunakan untuk menyimpan jumlah kemunculan setiap nilai intensitas piksel.

Selanjutnya, program menghitung histogram citra `merge` menggunakan dua buah perulangan `for` untuk mengakses setiap piksel pada gambar. Nilai piksel yang diperoleh disimpan pada variabel `pixel`, kemudian frekuensi kemunculan nilai tersebut ditambahkan pada array histogram menggunakan:
```python id="7a4vrx"
hist_asli[pixel] += 1

````python

hist_asli = np.zeros(256, dtype=int)
hist_gabungan = np.zeros(256, dtype=int)

for y in range(merge.shape[0]):
    for x in range(merge.shape[1]):

        pixel = merge[y][x]

        hist_asli[pixel] += 1

for y in range(nailong_sultan.shape[0]):
    for x in range(nailong_sultan.shape[1]):

        pixel = nailong_sultan[y][x]

        hist_gabungan[pixel] += 1


plt.figure(figsize=(10, 5))
plt.plot( hist_asli,label='New Image',linewidth=1.8)
plt.plot(hist_gabungan, label='Gabungan', linewidth=1.8)
plt.title( "Perbandingan Histogram",fontsize=15)
plt.xlabel("Nilai Pixel (0-255)",  fontsize=11)
plt.ylabel( "Frekuensi",fontsize=11)
plt.xlim(-12, 260)
plt.ylim(-12000, 255000)
plt.xticks( [0, 50, 100, 150, 200, 250], fontsize=10)
plt.yticks( [0, 50000, 100000, 150000, 200000, 250000], )
plt.legend( loc='upper right',  fontsize=10)
plt.grid(False)
plt.tight_layout()
plt.show()
````
